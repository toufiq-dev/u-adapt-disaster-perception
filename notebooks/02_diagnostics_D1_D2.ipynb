# 02 — Diagnostics D1/D2 (pilot)

**Purpose:** run the pre-registered D1/D2 checks **before** any main results are collected (proposal §11, pilot week):

- **D1** — text uncertainty (σ̃²_text) vs proposal-level error rate: 10 bins + Spearman ρ > 0 expected.
- **D2** — visual uncertainty (σ̃²_visual) vs visual-only error rate.

**If D1/D2 fail**, the pre-registered contingency plan applies (docs/pre_registration.md):
(a) report honestly and demote Mode A uncertainty terms to ablations;
(b) fall back to the plain-confidence baseline (cosine-similarity margin);
(c) rely on the Mode B logistic gate;
(d) pivot to the comparative calibration-and-reliability study.

```python
import numpy as np

from uadapt.metrics.diagnostics import (
    d1_text_uncertainty_accuracy,
    d2_visual_uncertainty_accuracy,
)

# TODO: real values flow from cached features + prototypes (Milestone 5).
rng = np.random.default_rng(0)
n = 1000
norm_text_var = rng.uniform(0, 1, n)
norm_visual_var = rng.uniform(0, 1, n)
correct = (rng.uniform(0, 1, n) > 0.3 + 0.5 * norm_text_var).astype(bool)

d1 = d1_text_uncertainty_accuracy(norm_text_var, correct)
d2 = d2_visual_uncertainty_accuracy(norm_visual_var, correct)
print("D1:", d1.summary, "|", d1.flag)
print("D2:", d2.summary, "|", d2.flag)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
for ax_, res, title in (
    (ax[0], d1, "D1: text uncertainty vs error rate"),
    (ax[1], d2, "D2: visual uncertainty vs error rate"),
):
    ax_.bar(res.raw["bin_mid"], res.raw["error_rate"], width=0.08)
    ax_.set_xlabel("normalized variance bin")
    ax_.set_ylabel("error rate")
    ax_.set_title(title)
plt.tight_layout()
plt.show()
```
